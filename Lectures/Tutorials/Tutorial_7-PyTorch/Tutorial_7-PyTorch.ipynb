**PyTorch** is an open-source machine learning library originally developed by Meta's AI research lab and released in 2016. Since 2022 it has been governed by the PyTorch Foundation, a part of the Linux Foundation. PyTorch provides tools for building neural networks, performing automatic differentiation, and running large-scale computations on CPUs and GPUs. Its core data structure is the **tensor**, a multi-dimensional array similar to a NumPy array, which makes PyTorch code feel familiar to Python and NumPy users.

Beyond the core library, the PyTorch ecosystem includes domain libraries such as **TorchVision** for computer vision and **TorchAudio** for audio processing, as well as a large collection of pretrained models and tools (e.g., the Hugging Face Transformers library is built on PyTorch). PyTorch is widely used in both research and production, and it is currently the dominant framework for deep learning research.

This tutorial introduces the basics of working with PyTorch tensors. Lecture 15 builds on these basics to define, train, and evaluate neural networks with PyTorch.

## 7.2 Tensor Basics

A **tensor** is a multi-dimensional array of numbers of a single data type. A scalar is a 0-dimensional tensor, a vector is a 1-dimensional tensor, a matrix is a 2-dimensional tensor, and so on. PyTorch tensors are very similar to NumPy arrays, with two important additions: they can be stored on a GPU, and they can record the operations applied to them for automatic differentiation.

<a id="7.2.1-creating-tensors"></a>

### 7.2.1 Creating Tensors

Tensors can be created from Python lists with `torch.tensor`, filled with constant values with `torch.zeros` or `torch.ones`, or filled with random numbers with `torch.rand` (uniform distribution on [0, 1)) and `torch.randn` (standard normal distribution).

````python
# Tensor from a nested Python list
x = torch.tensor([[1, 2, 3], [4, 5, 6]])

print(f"Type: {x.dtype}")
print(f"Size: {x.shape}")
print(f"Values: \n{x}")
````

````python
# Zeros and ones tensors
x = torch.zeros(2, 3)
print(x)

x = torch.ones(2, 3)
print(x)
````

````python
# Random tensor from a normal distribution
x = torch.randn(2, 3)

print(f"Type: {x.dtype}")
print(f"Size: {x.shape}")
print(f"Values: \n{x}")
````

````python
# Random tensor from a uniform distribution on [0, 1)
x = torch.rand(2, 3)
print(x)
````

The attribute `shape` (or equivalently the method `size()`) returns the dimensions of a tensor, `ndim` returns the number of dimensions, and `numel()` returns the total number of elements.

````python
x = torch.zeros(2, 3, 4)

print(f"Shape: {x.shape}")
print(f"Number of dimensions: {x.ndim}")
print(f"Number of elements: {x.numel()}")
````

<a id="7.2.2-tensor-data-types"></a>

### 7.2.2 Tensor Data Types

Each tensor has a data type `dtype`. The default type for floating-point tensors is 32-bit float (`torch.float32`), and the default type for integer tensors is 64-bit integer (`torch.int64`). Neural networks almost always use `torch.float32` tensors.

The `dtype` can be specified when creating a tensor, or changed afterward with the `to()` method (or with the shortcuts `float()`, `double()`, `int()`, `long()`).

````python
# Integer tensor (int64 by default)
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(f"Type: {x.dtype}")

# Specify the dtype at creation
x = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
print(f"Type: {x.dtype}")

# Convert to double precision (float64)
x_double = x.to(torch.float64)
print(f"Type: {x_double.dtype}")

# Convert to int32
x_int = x.to(torch.int32)
print(f"Type: {x_int.dtype}")
````

<a id="7.2.3-conversion-between-numpy-arrays-and-tensors"></a>

### 7.2.3 Conversion between NumPy Arrays and Tensors

A NumPy array can be converted into a tensor with `torch.from_numpy` or `torch.tensor`, and a tensor can be converted into a NumPy array with the `numpy()` method. Note that `torch.from_numpy` shares the memory with the original array (changing one changes the other), whereas `torch.tensor` always makes a copy of the data.

````python
# NumPy array to tensor
a = np.random.rand(2, 3)
x = torch.from_numpy(a)

print(f"Type: {x.dtype}")
print(f"Values: \n{x}")
````

````python
# Tensor to NumPy array
y = torch.ones(2, 3)
b = y.numpy()

print(type(b))
print(b)
````

<a id="7.3-operations"></a>

## 7.3 Operations

PyTorch provides hundreds of tensor operations, including arithmetic, linear algebra, reshaping, indexing, and reductions. Most of them mirror the corresponding NumPy functions.

<a id="7.3.1-arithmetic-and-matrix-operations"></a>

### 7.3.1 Arithmetic and Matrix Operations

Arithmetic operators (`+`, `-`, `*`, `/`) are applied element-wise, and follow the same broadcasting rules as NumPy.

````python
# Element-wise addition and multiplication
x = torch.tensor([[2., 2., 2.], [2., 2., 2.]])
y = torch.tensor([[3., 3., 3.], [3., 3., 3.]])

print(f"x + y = \n{x + y}")
print(f"x * y = \n{x * y}")
````

Matrix multiplication is performed with `torch.mm` for 2-dimensional tensors, or more generally with `torch.matmul` (also available as the `@` operator), which supports batched multiplication of higher-dimensional tensors.

````python
# Matrix multiplication
x = torch.tensor([[2., 2., 2.], [2., 2., 2.]])
y = torch.tensor([[3., 3.], [3., 3.], [3., 3.]])
z = torch.mm(x, y)

print(f"Size: {z.shape}")
print(f"Values: \n{z}")

# Equivalent with the @ operator
print(f"Values: \n{x @ y}")
````

````python
# Transpose
x = torch.tensor([[2., 2., 2.], [2., 2., 2.]])
print(f"Size: {x.shape}")
print(f"Values: \n{x}")

y = torch.t(x)  # or x.T
print(f"Size: {y.shape}")
print(f"Values: \n{y}")
````

<a id="7.3.2-reshaping"></a>

### 7.3.2 Reshaping

The method `view` returns a tensor with the same data but a different shape, and `reshape` does the same but it also works for tensors that are not stored contiguously in memory. A dimension of `-1` is inferred from the remaining dimensions.

````python
x = torch.tensor([[2., 2., 2.], [2., 2., 2.]])

z = x.view(3, 2)
print(f"Size: {z.shape}")
print(f"Values: \n{z}")

# -1 infers the size of that dimension
z = x.reshape(-1)
print(f"Size: {z.shape}")
print(f"Values: \n{z}")
````

`unsqueeze` adds a new dimension of size 1 at a given position, and `squeeze` removes dimensions of size 1. These are commonly used to add or remove a batch dimension.

````python
x = torch.tensor([1., 2., 3.])
print(f"Size: {x.shape}")

y = x.unsqueeze(0)  # add a batch dimension
print(f"Size: {y.shape}, Values: {y}")

z = y.squeeze(0)    # remove the batch dimension
print(f"Size: {z.shape}, Values: {z}")
````

<a id="7.3.3-reductions"></a>

### 7.3.3 Reductions

Reduction operations such as `sum`, `mean`, `max`, and `argmax` can be applied to the whole tensor, or along a specified dimension `dim`. Note that `dim=0` reduces over the rows (producing one value per column), and `dim=1` reduces over the columns (producing one value per row).

````python
x = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print(f"Values: \n{x}")

print(f"Sum of all elements: {torch.sum(x)}")
print(f"Sum over dim=0 (per column): {torch.sum(x, dim=0)}")
print(f"Sum over dim=1 (per row): {torch.sum(x, dim=1)}")
print(f"Mean over dim=1 (per row): {x.mean(dim=1)}")
print(f"Index of the max in each row: {x.argmax(dim=1)}")
````

<a id="7.3.4-indexing,-slicing,-and-joining"></a>

### 7.3.4 Indexing, Slicing, and Joining

Tensors support the same indexing and slicing syntax as NumPy arrays.

````python
x = torch.tensor([[1., 2., 3.], [4., 5., 6.], [7., 8., 10.], [11., 12., 13.]])
print(f"x: \n{x}")

# First row
print(f"x[0]: {x[0]}")

# Last column
print(f"x[:, -1]: {x[:, -1]}")

# First and second rows, second and third columns
print(f"x[:2, 1:3]: \n{x[:2, 1:3]}")
````

Boolean masks can be used to select the elements that satisfy a condition.

````python
# Boolean indexing
mask = x > 6
print(f"Mask: \n{mask}")
print(f"x[x > 6]: {x[mask]}")
````

Tensors are joined along an existing dimension with `torch.cat`, or along a new dimension with `torch.stack`.

````python
x = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print(f"Values: \n{x}")

y = torch.cat([x, x], dim=0)  # join by rows
print(f"cat along dim=0, size {y.shape}: \n{y}")

y = torch.cat([x, x], dim=1)  # join by columns
print(f"cat along dim=1, size {y.shape}: \n{y}")

z = torch.stack([x, x])       # new leading dimension
print(f"stack, size {z.shape}")
````

<a id="7.4-gradients"></a>

## 7.4 Gradients

Training neural networks relies on computing gradients of a loss function with respect to the model parameters. PyTorch's automatic differentiation engine, **autograd**, does this for us. When a tensor is created with `requires_grad=True`, PyTorch records all operations applied to it in a computational graph. Calling `backward()` on a scalar output then traverses this graph in reverse and accumulates the gradients in the `grad` attribute of the input tensors.

For example, for

* $x = 3$
* $y = x^2 + 2x + 1$

the derivative is $\frac{dy}{dx} = 2x + 2 = 2 \cdot 3 + 2 = 8$.

````python
# Create a tensor with gradient tracking
x = torch.tensor([3.0], requires_grad=True)

# Define a simple quadratic function
y = x**2 + 2*x + 1

# backward() requires a scalar output, so we sum the elements of y
y = y.sum()

# Backpropagate to calculate dy/dx = 2x + 2
y.backward()

# Evaluate the derivative dy/dx for x = 3, i.e., dy/dx = 2*3 + 2 = 8
print(f"x: {x}")
print(f"Gradient: {x.grad}")
````

Gradients also work with vector inputs. In the next example, $y = \sum_i x_i^2$, so the gradient with respect to each element is $2x_i$. Note that gradients are accumulated in `grad` each time `backward()` is called, which is why `zero_grad()` is called before each training step in a typical PyTorch training loop.

````python
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y = (x**2).sum()
y.backward()

print(f"x: {x}")
print(f"Gradient: {x.grad}")
````

When we only need to evaluate a model (e.g., at inference time), gradient tracking is unnecessary and it can be switched off with the `torch.no_grad()` context manager, or by calling `detach()` on a tensor.

````python
with torch.no_grad():
    z = x * 2
print(f"requires_grad inside no_grad(): {z.requires_grad}")

w = x.detach()
print(f"requires_grad after detach(): {w.requires_grad}")
````

## 7.5 CUDA Tensors

**CUDA** is NVIDIA's platform for running computations on GPUs. PyTorch can move tensors and models to a GPU, where operations on large tensors are much faster than on a CPU. First, we check whether a CUDA-capable GPU is available.

````python
# Is CUDA available?
print(torch.cuda.is_available())
````

If the output is `False`, a GPU is not available. In Google Colab, we can switch to a GPU runtime via *Runtime* > *Change runtime type* > *Hardware accelerator* > *T4 GPU*.

A common pattern is to define a `device` object that selects the GPU when it is available and falls back to the CPU otherwise, so that the same code runs in both settings.

````python
# Set the device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
````

````python
# Move a tensor to the device
x = torch.rand(2, 3).to(device)
print(f"Device: {x.device}")
print(f"Is CUDA tensor: {x.is_cuda}")
````

Operations can only be performed between tensors on the same device. A tensor on the GPU must be moved back to the CPU with `cpu()` before converting it to a NumPy array.

````python
y = torch.rand(2, 3).to(device)
z = x + y            # both tensors are on the same device
print(z.cpu().numpy())
````
